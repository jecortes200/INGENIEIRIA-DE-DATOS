```python
import pandas as pd

df = pd.read_csv('ANEEL_Dataset.csv')

# Vemos cuántas filas y columnas tiene, y las primeras 5 filas
print(f"Filas: {df.shape[0]} | Columnas: {df.shape[1]}")
df.head()
```

```python
# Quitar espacios al inicio o final de los nombres de las columnas
df.columns = df.columns.str.strip()

df = df.drop_duplicates()

print("Nombres de columnas limpios:")
print(df.columns.tolist())
```

```python
# cambiamos ',' por '.'
cols_decimales = ['pot_MW', 'garantia_fisica_MW_medio', 'preco_por_MW']
for col in cols_decimales:
    df[col] = df[col].astype(str).str.replace(',', '.')
    df[col] = pd.to_numeric(df[col], errors='coerce') # Convertir a número real

df['invest_previsto'] = df['invest_previsto'].astype(str).str.replace('.', '', regex=False).str.replace(',', '.', regex=False)
df['invest_previsto'] = pd.to_numeric(df['invest_previsto'], errors='coerce')

df.info()
```

```python
df = df.dropna(subset=['pot_MW', 'invest_previsto'])

# Crear una nueva columna calculada: ¿Cuánta plata se invirtió por cada MW?
df['invest_por_MW'] = df['invest_previsto'] / df['pot_MW']

df.head()
```

```python
df.to_csv('ANEEL_Dataset_Procesado.csv', index=False)

print("¡Corona! El dataset limpio y está como 'ANEEL_Dataset_Procesado.csv'")
```
